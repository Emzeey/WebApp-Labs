1. Napisz program, który z serwera ntp.task.gda.pl pobierze aktualną datę i czas, a następnie wyświetli je na konsoli. Serwer działa na porcie 13.

```python
def ntp():
    host = "ntp.task.gda.pl"
    port = 13

    try:
        with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as sock:
            sock.settimeout(5.0)
            sock.connect((host, port))
            recv_data = sock.recv(1024)
            print(f"Date and time: {recv_data.decode('utf-8')}")
    except Exception as e:
        print(f"Error: {e}")

ntp()
```

7. Zmodyfikuj program numer 6 z laboratorium nr 1 w ten sposób, aby oprócz wyświetlania informacji o
tym, czy port jest zamknięty, czy otwarty, klient wyświetlał również informację o tym, jaka usługa jest
uruchomiona na danym porcie.

```python
def connect_by(ip: str, port: int) -> None:
    try:
        service_name = socket.getservbyport(port, "tcp")
    except OSError:
        service_name = "Unknown service"

    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.settimeout(3)
        try:
            sock.connect((ip, port))
            print(f"Port {port} on {ip} is OPEN (Service: {service_name})")
        except Exception as e:
            print(f"Port {port} on {ip} is CLOSED (Service: {service_name})")
            print(e)

connect_by("127.0.0.1", 8888)
connect_by("8.8.8.8", 8888)
connect_by("8.8.8.8", 53)
connect_by("127.0.0.1", 80)
```

8. Zmodyfikuj program numer 7 z laboratorium nr 1 w ten sposób, aby oprócz wyświetlania informacji o
tym, czy porty są, jest zamknięte, czy otwarte, klient wyświetlał również informację o tym, jaka usługa
jest uruchomiona na danym porcie.

```python
def port_scanner(start_port: int, end_port: int):
    host = input("IP address: ")
    print(f"Scanning {host} from port {start_port} to {end_port}...\n")

    for port in range(start_port, end_port + 1):
        with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
            sock.settimeout(0.5)
            result = sock.connect_ex((host, port))

            try:
                service_name = socket.getservbyport(port, "tcp")
            except OSError:
                service_name = "Unknown service"

            if result == 0:
                print(f"Port {port} is OPEN (Service: {service_name})")
            else:
                print(f"Port {port} is CLOSED (Service: {service_name})")

    print("\nScanning complete")

start_port_input = int(input("Start port: "))
end_port_input = int(input("End port: "))
port_scanner(start_port_input, end_port_input)
```

11. Zmodyfikuj program nr 2 z laboratorium nr 2 w ten sposób, aby klient wysłał i odebrał od serwera wiadomość o maksymalnej długości 20 znaków. Serwer TCP odbierający i wysyłający wiadomości o długości 20 działa pod adresem 212.182.24.27 na porcie 2908. Uwzględnij sytuacje, gdy:
- wiadomość do wysłania jest za krótka -> ma być wówczas uzupełniania do 20 znaków znakami spacji
- wiadomość do wysłania jest za długa -> ma być przycięta do 20 znaków (lub wysłana w całości -> sprawdź, co się wówczas stanie)

```python
sockIPv4 = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

try:
    sockIPv4.connect(('127.0.0.1', 2908))
    user_input = input("Message: ")
    MAX_PACKET_LENGTH = 20
    if len(user_input) < MAX_PACKET_LENGTH:
        user_input = user_input.ljust(MAX_PACKET_LENGTH)
    else:
        user_input = user_input[:MAX_PACKET_LENGTH]
    sockIPv4.send(user_input.encode())
    data = sockIPv4.recv(4096)
    print(data.decode())
except socket.error as exc:
    print(f"Error {exc}")

sockIPv4.close()
```

12. Funkcje recv i send nie gwarantują wysłania / odbioru wszystkich danych. Rozważmy funkcję recv. Przykładowo, 100 bajtów może zostać wysłane jako grupa po 10 bajtów, albo od razu w całości. Oznacza to, iż jeśli używamy gniazd TCP, musimy odbierać dane, dopóki nie mamy pewności, że odebraliśmy ich ilość. Zmodyfikuj program nr 11 z laboratorium nr 2 w ten sposób, aby mieć pewność, że klient w rzeczywistości odebrał / wysłał wiadomość o wymaganej długości.

```python
def recall(sock, msg_len):
    msg = b""
    bytes_rcv = 0
    while bytes_rcv < msg_len:
        chunk = sock.recv(msg_len - bytes_rcv)
        if not chunk:
            break
        msg += chunk
        bytes_rcv += len(chunk)
    return msg

def sendall_fixed(sock, data):
    total_sent = 0
    while total_sent < len(data):
        sent = sock.send(data[total_sent:])
        if sent == 0:
            raise RuntimeError("socket connection broken")
        total_sent += sent

sockIPv4 = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

try:
    sockIPv4.connect(('127.0.0.1', 2908))
    user_input = input("Message: ")
    MAX_PACKET_LENGTH = 20
    if len(user_input) < MAX_PACKET_LENGTH:
        user_input = user_input.ljust(MAX_PACKET_LENGTH)
    else:
        user_input = user_input[:MAX_PACKET_LENGTH]
    data_bytes = user_input.encode()
    sendall_fixed(sockIPv4, data_bytes)
    received_bytes = recall(sockIPv4, MAX_PACKET_LENGTH)
    print(received_bytes.decode())
except socket.error as exc:
    print(f"Error {exc}")

sockIPv4.close()
```
