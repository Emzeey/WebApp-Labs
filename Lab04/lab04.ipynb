1. Napisz program serwera, który działając pod adresem 127.0.0.1 oraz na określonym porcie TCP, dla
podłączającego się klienta, będzie odsyłał mu aktualny czas oraz datę. Prawidłowa komunikacja powinna
odbywać się w następujący sposób:
- Serwer odbiera od klienta wiadomość (dowolną)
- Serwer odsyła klientowi aktualną datę i czas

```python
HOST = '127.0.0.1'
PORT = 3000

if __name__ == "__main__":
    sockIPv4 = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    try:
        sockIPv4.connect((HOST, PORT))

        sockIPv4.send(b"SMH")
        print(sockIPv4.recv(1024).decode())
    except socket.error as exc:
        print(f"Error {exc}")

    print("Closing")
    sockIPv4.close()
```

7. Zmodyfikuj program nr 2 z laboratorium nr 3 w ten sposób, aby serwer wysyłał i odbierał wiadomość o maksymalnej długości 20 znaków.

```python
import socket

HOST = '127.0.0.1'
PORT = 3000
MAX_LEN = 20

if __name__ == "__main__":
    sockIPv4 = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    try:
        sockIPv4.connect((HOST, PORT))

        message = "Testsdfasdfasfd123456789"

        if len(message) > MAX_LEN:
            print(f"WARNING: message too long ({len(message)} chars), truncating to {MAX_LEN}.")
            message = message[:MAX_LEN]

        sockIPv4.send(message.encode())
        print(f"Sent: \"{message}\"")

        response = sockIPv4.recv(MAX_LEN).decode()
        print(f"Received: \"{response}\"")

    except socket.error as exc:
        print(f"Error: {exc}")

    print("Closing")
    sockIPv4.close()
```

11. Napisz program serwera, który działając pod adresem 127.0.0.1 oraz na określonym porcie UDP, dla podłączającego się klienta, odbierze od niego wiadomość o treści podanej w zadaniu nr 15 z laboratorium nr 3, a następnie odeśle klientowi odpowiedź TAK, lub NIE. W przypadku błędnego sformatowania wiadomości, serwer odeśle klientowi odpowiedź BAD SYNTAX.

```python
import socket

SERVER_HOST = '127.0.0.1'
SERVER_PORT = 2911

IP_RAW = bytes.fromhex(
    "45000 04ef7fa40003806 9d33d4b6181bc0a80002"
    "0b54b9a6fbf93c57c10a06c1801800e3ce9c00000101080a03a6eb01000bf8e5"
    "6e6574776f726b2070726f6772616d6d696e6720697320 66756e".replace(" ", "")
)

ip_version = IP_RAW[0] >> 4
ip_ihl     = (IP_RAW[0] & 0x0F) * 4
ip_proto   = IP_RAW[9]
src_ip     = '.'.join(str(b) for b in IP_RAW[12:16])
dst_ip     = '.'.join(str(b) for b in IP_RAW[16:20])

tcp_start  = ip_ihl
src_port   = int.from_bytes(IP_RAW[tcp_start:tcp_start+2], 'big')
dst_port   = int.from_bytes(IP_RAW[tcp_start+2:tcp_start+4], 'big')
tcp_offset = (IP_RAW[tcp_start+12] >> 4) * 4
data       = IP_RAW[tcp_start + tcp_offset:].decode()

with socket.socket(socket.AF_INET, socket.SOCK_DGRAM) as sock:
    msg_a = f"zad15odpA;ver;{ip_version};srcip;{src_ip};dstip;{dst_ip};type;{ip_proto}"
    print(f"Sending A: {msg_a}")
    sock.sendto(msg_a.encode(), (SERVER_HOST, SERVER_PORT))

    response_a, _ = sock.recvfrom(1024)
    print(f"Response A: {response_a.decode()}")

    if response_a.decode().strip() == "TAK":
        msg_b = f"zad15odpB;srcport;{src_port};dstport;{dst_port};data;{data}"
        print(f"Sending B: {msg_b}")
        sock.sendto(msg_b.encode(), (SERVER_HOST, SERVER_PORT))

        response_b, _ = sock.recvfrom(1024)
        print(f"Response B: {response_b.decode()}")
    else:
        print("Skipping step B — step A was incorrect.")
```
